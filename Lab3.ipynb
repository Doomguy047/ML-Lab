### **Aim:** To perform statistical and correlation analysis on the dataset to identify relationships between features and the target class.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
df = pd.read_csv("/Users/namanjain/Downloads/ML LAB/ML_Labs/Heart_Disease_Prediction.csv")
```

```python
df.mean(numeric_only=True)
```

```python
df.median(numeric_only=True)
```

```python
df.std(numeric_only=True)
```

```python
df.corr(numeric_only=True)
```

```python
plt.figure(figsize=(10, 6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.show()
```

```python
numeric_df = df.select_dtypes(include=np.number)
```

```python
sns.scatterplot(x=numeric_df.iloc[:, 0], y=numeric_df.iloc[:, 1])
plt.show()
```

```python
sns.boxplot(data=numeric_df)
plt.show()
```

```python
sns.violinplot(data=numeric_df)
plt.show()
```
